````python
# import os
# import json
# from dotenv import load_dotenv
# from langchain_pinecone import PineconeVectorStore
# from langchain_google_genai import GoogleGenerativeAIEmbeddings, ChatGoogleGenerativeAI
import os
import json
from dotenv import load_dotenv
from langchain_pinecone import PineconeVectorStore
from langchain_nvidia_ai_endpoints import NVIDIAEmbeddings
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_google_genai import ChatGoogleGenerativeAI


load_dotenv()

# ----------------------------
# 1. Initialize components
# ----------------------------
embeddings = NVIDIAEmbeddings(
    model="nvidia/nemotron-3-embed-1b:free",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1"
)

vectorstore = PineconeVectorStore(
    index_name="ecommerce-products",
    embedding=embeddings
)

llm = ChatGoogleGenerativeAI(
    model="gemini-3-flash-preview",        # or your preferred model
    temperature=0.1,
    google_api_key=os.getenv("GOOGLE_API_KEY")
)

# 2. Intelligent Router & Dynamic Query Analyzer
def analyze_and_route_query(user_query: str) -> dict:
    """
    Analyzes the user query:
    - Automatically converts any dynamic price threshold (e.g. 10k, 30000, 80000), 
      RAM size, or PTA request into Pinecone metadata filters.
    - Rewrites and expands general queries into technical search keywords.
    """
    router_prompt =  f"""You are a precise query parser for an e-commerce vector database (Pinecone).

Exact Metadata Schema & Data Types:
- `price_numeric` (int in PKR): Current selling price (e.g., 30000, 150000). Convert shorthand ("30k" -> 30000, "1.5 lakh" -> 150000).
- `original_price_numeric` (int in PKR): Original price before discount.
- `discount_pct` (int): Discount percentage (e.g., 10, 20).
- `pta_status` (bool): `true` if PTA approved/official, `false` if non-PTA/unapproved.
- `ram_gb` (int): RAM in GB (e.g., 4, 8, 12, 16).
- `storage_gb` (int): Storage in GB. Convert units ("1TB" -> 1000 or 1024, "512GB" -> 512).
- `display_size_inches` (float): Screen size in inches (e.g., 6.5, 6.78).
- `battery_mah` (int): Battery capacity in mAh (e.g., 5000, 6000).
- `main_camera_mp` (int): Rear camera resolution in MP (e.g., 50, 108).
- `color_names` (string): Color name in lowercase (e.g., "black", "blue", "white").
- `product_name` (string): Model or brand name string.

(Note: `url`, `color_images`, `whats_in_box_str`, and `color_statuses` are display fields—do not filter on them.)

User Query: "{user_query}"

Instructions:
1. Extract ALL explicit filter constraints from the query into `pinecone_filter`.
2. Operators to use: `$lte`, `$gte`, `$eq`.
3. Range Queries (e.g., "6.5 to 6.8 inch display", "100k to 300k price", "8 to 16GB RAM"):
   Combine `$gte` and `$lte` inside the field condition.
4. Multiple Conditions: Enclose ALL conditions inside a single top-level `"$and"` array.
5. Clean `search_query`: Strip specific numbers already captured in filters to keep vector search semantic.

Few-Shot Examples:

- Example 1 (Display Size & Discount):
  Query: "Phones with at least 6.5 inch screen and more than 10% discount under 50k"
  JSON: {{
    "has_hardcoded_specs": true,
    "pinecone_filter": {{
      "$and": [
        {{"display_size_inches": {{"$gte": 6.5}}}},
        {{"discount_pct": {{"$gte": 10}}}},
        {{"price_numeric": {{"$lte": 50000}}}}
      ]
    }},
    "search_query": "large display smartphone on sale"
  }}

- Example 2 (Storage Range in TB, RAM, PTA):
  Query: "PTA approved phone with 128gb to 1tb storage and 8gb to 16gb ram"
  JSON: {{
    "has_hardcoded_specs": true,
    "pinecone_filter": {{
      "$and": [
        {{"pta_status": true}},
        {{"storage_gb": {{"$gte": 128}}}},
        {{"storage_gb": {{"$lte": 1000}}}},
        {{"ram_gb": {{"$gte": 8}}}},
        {{"ram_gb": {{"$lte": 16}}}}
      ]
    }},
    "search_query": "flagship phone"
  }}

- Example 3 (Camera, Battery, Color):
  Query: "50MP main camera with 6000mah battery in black color"
  JSON: {{
    "has_hardcoded_specs": true,
    "pinecone_filter": {{
      "$and": [
        {{"main_camera_mp": {{"$gte": 50}}}},
        {{"battery_mah": {{"$gte": 6000}}}},
        {{"color_names": {{"$eq": "black"}}}}
      ]
    }},
    "search_query": "high camera battery smartphone"
  }}

- Example 4 (Pure Conceptual - No Hardcoded Specs):
  Query: "I need a simple keypad phone for my elderly father"
  JSON: {{
    "has_hardcoded_specs": false,
    "pinecone_filter": null,
    "search_query": "basic keypad feature phone physical buttons simple operation"
  }}

Return ONLY a valid JSON object matching this structure:
{{
  "has_hardcoded_specs": true/false,
  "pinecone_filter": dict or null,
  "search_query": "string"
}}

JSON Output ONLY:"""

    response = llm.invoke(router_prompt)
    
    # Extract raw text response
    raw_text = response.content
    if isinstance(raw_text, list):
        raw_text = "".join(b.get("text", "") for b in raw_text if isinstance(b, dict) and "text" in b)
    
    clean_json_str = raw_text.strip().replace("```json", "").replace("```", "").strip()
    
    try:
        return json.loads(clean_json_str)
    except Exception:
        return {
            "has_hardcoded_specs": False,
            "pinecone_filter": None,
            "search_query": user_query
        }

# 3. Main RAG Pipeline with Real-Time Streaming & UI Data Preservation
def run_streaming_rag(user_query: str, top_k: int = 10):
    print(f"\n🔍 Original User Query: '{user_query}'")
    print("--------------------------------------------------")
    
    # Step A: Query Routing & Rewriting
    analysis = analyze_and_route_query(user_query)
    
    search_query = analysis.get("search_query", user_query)
    pinecone_filter = analysis.get("pinecone_filter")
    
    if analysis.get("has_hardcoded_specs") and pinecone_filter:
        print(f"🎯 Dynamic Specs Filter Detected: {pinecone_filter}")
    else:
        print(f"✨ Enhanced/Rewritten Search Query: '{search_query}'")

    # Step B: Vector Search with Full Metadata Retrieval
    print(f"📦 Fetching top {top_k} matching documents from Pinecone...")
    
    if pinecone_filter:
        retrieved_docs = vectorstore.similarity_search(search_query, k=top_k, filter=pinecone_filter)
    else:
        retrieved_docs = vectorstore.similarity_search(search_query, k=top_k)

    if not retrieved_docs:
        print("\n❌ No documents found matching your filter criteria.")
        return []

    print(f"✅ Found {len(retrieved_docs)} matching chunks.")
    print("--------------------------------------------------\n🤖 AI Response (Streaming):\n")

    # Step C: Format All Metadata & Content into Context
    context = "\n\n---\n\n".join(
        f"Metadata: {doc.metadata}\nContent: {doc.page_content}"
        for doc in retrieved_docs
    )

    final_prompt = f"""You are an expert E-Commerce Sales & Technical Assistant.
Answer the customer's question using ONLY the provided product catalog context below.
Include matching products with their specs, price, PTA status, and box contents.

Product Catalog Context:
{context}

Customer Question:
{user_query}

Helpful Answer:"""

    # Step D: Real-Time Character/Token Streaming
    stream = llm.stream(final_prompt)
    for chunk in stream:
        content = chunk.content
        if isinstance(content, str):
            print(content, end="", flush=True)
        elif isinstance(content, list):
            for block in content:
                if isinstance(block, dict) and "text" in block:
                    print(block["text"], end="", flush=True)
                elif isinstance(block, str):
                    print(block, end="", flush=True)

    print("\n\n--------------------------------------------------")
    
    # Returns docs so Streamlit can render product link buttons & images from doc.metadata
    return retrieved_docs

# 4. Execution Tests
if __name__ == "__main__":
    # Test 1: Dynamic price (under 30000)
    retrieved = run_streaming_rag("Show me PTA approved phones between 100000 to 300000 rupees")
    
    # Test 2: Dynamic price (under 80000) & RAM spec
    retrieved = run_streaming_rag("I need a phone for myself as i am unliteral and can't operate smartphone")
````
