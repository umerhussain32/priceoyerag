```python
import os
from dotenv import load_dotenv
import pandas as pd
from langchain_core.documents import Document
from langchain_pinecone import PineconeVectorStore
from langchain_nvidia_ai_endpoints import NVIDIAEmbeddings


print("All set. You are ready to go!")
```

### Loading Cleaned Dataset

```python
df=pd.read_csv("cleaned_dataset_rag.csv")

df
```

### Metadata configuration for Documents & Chunking
  MetaData will help in searching hard facts like price, urls etc. through sql queries

```python
all_metadata=[]
all_documents=[]
for i, row in df.iterrows():
   metadata={
       "product_name": str(row.get("product_name","")),
       "url": str(row.get("url","")),
       "whats_in_box_str": str(row.get("whats_in_the_box", "")),
       "color_statuses": str(row.get("color_statuses", "")),
       "color_images": str(row.get("color_images","")),
       "color_names": str(row.get("color_names","")),
       "price_numeric": int(row.get("price_numeric",0)),
       "original_price_numeric": int(row.get("original_price_numeric",0)),
       "discount_pct": int(row.get("discount_pct",0)),
       "battery_mah":int(row.get("battery_mah",0)),
       "ram_gb":int(row.get("ram_gb",0)),
       "storage_gb":int(row.get("storage_gb",0)),
       "display_size_inches": float(row.get("display_size_inches",0.0)),
       "main_camera_mp":int(row.get("main_camera_mp",0)),
       "pta_status": "approved" in str(row.get("pta_status","")).lower()
   }
   all_metadata.append(metadata)

   specs_text = f"""
Product Name: {row.get('product_name', 'N/A')}
Current Price: {row.get('current_price_str', 'N/A')}
Original Price: {row.get('original_price_str', 'N/A')}
Discount: {row.get('discount_str', 'N/A')}
PTA Status: {row.get('pta_status', 'N/A')}
Display Specs: {row.get('display', 'N/A')}
Processor: {row.get('processor', 'N/A')}
Memory & Storage: {row.get('memory', 'N/A')}
Battery Specs: {row.get('battery', 'N/A')}
Camera Specs: {row.get('camera', 'N/A')}
Available Colors: {row.get('color_names', 'N/A')}
Color Stock Status: {row.get('color_statuses', 'N/A')}
Product Images: {row.get('color_images', 'N/A')}
What's In The Box: {row.get('whats_in_the_box', 'N/A')}
Warranty & Delivery: {row.get('warranty_terms', 'N/A')} | {row.get('delivery_badges', 'N/A')}
""".strip()

   specs_doc = Document(
        page_content=specs_text,
        metadata={**metadata, "chunk_type": "specs_text"}
    )
   all_documents.append(specs_doc)

   social_text=f"""
Product Name:{row.get("product_name","N/A")}
Frequently Asked Questions:{row.get("faqs_text","No FAQ Available")}
Reviews & Ratings:{row.get("reviews_text","No Reviews & Ratings Available")}
"""
   social_doc=Document(
        page_content=social_text,
        metadata={**metadata,"chunk_type":"social_text"}
            )
   all_documents.append(social_doc)



for i in range(0,33):
    print(all_documents[i],"\n")

print(f"Total documents created: {len(all_documents)}")
```

### Embedding

```python
load_dotenv()
embeddings = NVIDIAEmbeddings(
    model="nvidia/nemotron-3-embed-1b:free",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1"
)

index = "ecommerce-products"

vectorstore=PineconeVectorStore.from_documents(
    documents=all_documents,
    embedding=embeddings,
    index_name=index
    )
print("Embedding storing finished")
```

```python
# Test retrieving uploaded chunks
results = vectorstore.similarity_search("Oppo phone", k=2)

for doc in results:
    print("Found Document:", doc.page_content[:1000])
```

```python
# Query with metadata filtering (e.g., PTA approved & under 50k)
results = vectorstore.similarity_search(
    query="phones with great camera and long battery life",
    k=3,
    filter={
        "pta_status": True,
        "price_numeric": {"$lte": 50000}
    }
)

for doc in results:
    print(f"Product: {doc.metadata.get('product_name')} | Price: {doc.metadata.get('price_numeric')}")
```
